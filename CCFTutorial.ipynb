# Join cell_metadata with ccf_coordinates

```python
cell_metadata.columns
```

```python
cell_ccf_coordinates.rename(
    {'x': 'x_ccf',
     'y': 'y_ccf',
     'z': 'z_ccf'},
    axis=1,
    inplace=True
)
```

```python
cell_metadata = cell_metadata.join(
    cell_ccf_coordinates.set_index('cell_label'),
    on='cell_label'
)
```

```python
cell_metadata = cell_metadata.dropna()
```

# Download CCF

```python
ccf_dir = 'Allen-CCF-2020'
```

```python
abc_cache.list_metadata_files(directory=ccf_dir)
```

```python
parcellation_term_set = abc_cache.get_metadata_dataframe(
    directory=ccf_dir,
    file_name='parcellation_term_set'
)
```

```python
parcellation_term_set
```

```python
membership = abc_cache.get_metadata_dataframe(
    directory=ccf_dir,
    file_name='parcellation_to_parcellation_term_membership'
)
```

```python
membership
```

```python
for term_set_label in membership.parcellation_term_set_label.unique():
    subset = membership[membership.parcellation_term_set_label == term_set_label]
    term_set_name = subset.parcellation_term_set_name.iloc[0]
    subset = subset[['parcellation_index', 'parcellation_term_name', 'parcellation_term_acronym']]
    subset = subset.rename(
        {'parcellation_term_name': f'{term_set_name}_name',
         'parcellation_term_acronym': f'{term_set_name}_acronym'},
        axis=1
    )
    cell_metadata = cell_metadata.join(
        subset.set_index('parcellation_index'),
        on='parcellation_index'
    )
    print(f"joined '{term_set_name}'")
    
```

```python
cell_metadata
```

# Compare CA1 and CA3

```python
ca1 = cell_metadata[cell_metadata.structure_acronym=='CA1']
```

```python
ca3 = cell_metadata[cell_metadata.structure_acronym=='CA3']
```

```python
import anndata
```

```python
abc_cache.list_expression_matrix_files(directory=merfish_dir)
```

```python
merfish_expression_path = abc_cache.get_data_path(
    directory=merfish_dir,
    file_name='C57BL6J-638850/log2'
)
```

```python
merfish_expression = anndata.read_h5ad(
    merfish_expression_path,
    backed='r'
)
```

```python
merfish_expression
```

```python
ca1_data = merfish_expression[merfish_expression.obs.index.isin(ca1.cell_label)].to_memory()
```

```python
ca3_data = merfish_expression[merfish_expression.obs.index.isin(ca3.cell_label)].to_memory()
```

```python
len(ca1_data)
```

```python
len(ca3_data)
```

```python
import scipy.stats
```

```python
wilcoxon = scipy.stats.ranksums(ca1_data.X.toarray(), ca3_data.X.toarray(), axis=0)
```

```python
delta = (ca1_data.X.mean(axis=0)-ca3_data.X.mean(axis=0))
```

```python
delta.shape
```

```python
delta = np.array(delta).flatten()
```

```python
delta.shape
```

```python
delta.max()
```

```python
valid = np.logical_and(
    wilcoxon.pvalue<0.01,
    delta>1.0
)
```

```python
valid_idx = np.where(valid)
```

```python
valid_idx
```

```python
delta[valid_idx]
```
